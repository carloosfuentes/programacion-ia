```python
# Importando MNIST
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
X = digits.data
y = digits.target

print(X.shape)
print(y.shape)
```

```python
# Importando librerias de KMeans
from sklearn.cluster import KMeans

# Creando el modelo de KMeans con 10 clusters
kmeans = KMeans(n_clusters=10, random_state=42)

# Entrenando el modelo
y_pred = kmeans.fit_predict(X)

# Obteniendo las etiquetas despues de haber hecho el fit_predict
labels = kmeans.labels_

# Después de obtener las etiquetas, agrega al DataFrame
digits['cluster'] = labels
print(labels)
```

```python
# Los centroides son datos de 784 valores. Pinta los 10 centroides como si fueran imágenes.
import matplotlib.pyplot as plt
centroids = kmeans.cluster_centers_
plt.figure(figsize=(20, 10))
i = 0

for centroid in centroids:
    some_digit_image = centroid.reshape(8, 8)
    plt.subplot(1,10, i+1)
    plt.imshow(some_digit_image, cmap='binary')
    plt.title(f'Centroide {i}')
    plt.axis('off')
    i += 1

plt.show()
```
